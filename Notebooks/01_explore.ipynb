```python
# 1. Are any dates missing from train? (Dec 25 is known to be absent each year)
all_days = pd.date_range(train["date"].min(), train["date"].max())
missing_days = all_days.difference(train["date"].unique())
print(len(missing_days), missing_days)

# 2. Negative or strange sales?
print((train["sales"] < 0).sum(), "negative sales rows")
print((train["sales"] == 0).mean(), "fraction of zero-sales rows")

# 3. Do all store numbers in train exist in stores?
print(set(train["store_nbr"]) - set(stores["store_nbr"]))

# 4. Holidays: several rows on the same date?
print(holidays["date"].duplicated().sum(), "dates with more than one holiday row")
print(holidays["type"].value_counts())
print(holidays["transferred"].value_counts())
```
